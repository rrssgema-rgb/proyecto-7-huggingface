```python

import pandas as pd
import random
import json
import os

# Necesitamos las plantillas y respuestas de la celda anterior.
if "plantillas" not in globals() or "respuestas" not in globals():
    raise RuntimeError(
        "No encuentro las variables plantillas y respuestas. "
        "Vuelve a ejecutar la celda donde las definimos."
    )

random.seed(42)

grupos = {}

for categoria, casos in plantillas.items():
    lista = []

    for numero, (consulta, prioridad) in enumerate(casos):
        variantes = [
            consulta,
            "Buenos días. " + consulta,
            consulta + " Gracias.",
            "Necesito ayuda: " + consulta[0].lower() + consulta[1:],
            "Por favor, " + consulta[0].lower() + consulta[1:],
        ]

        ejemplos = []

        for texto in variantes:
            salida = {
                "category": categoria,
                "priority": prioridad,
                "response": respuestas[categoria],
            }

            ejemplos.append({
                "text": texto,
                "category": categoria,
                "priority": prioridad,
                "response": respuestas[categoria],
                "output_json": json.dumps(
                    salida, ensure_ascii=False
                ),
            })

        lista.append({
            "group_id": f"{categoria}_{numero:02d}",
            "examples": ejemplos,
        })

    random.shuffle(lista)
    grupos[categoria] = lista

train_rows = []
validation_rows = []
test_rows = []

# Seis consultas originales por categoría para entrenamiento.
# Validación y prueba reciben grupos completos, nunca variantes sueltas.
for categoria, lista in grupos.items():
    train_groups = lista[:6]
    remaining = lista[6:]

    for grupo in train_groups:
        for ejemplo in grupo["examples"]:
            train_rows.append({
                **ejemplo,
                "group_id": grupo["group_id"],
            })

    # Una consulta original por categoría para cada partición.
    validation_group = remaining[0]
    test_group = remaining[1]

    for ejemplo in validation_group["examples"]:
        validation_rows.append({
            **ejemplo,
            "group_id": validation_group["group_id"],
        })

    for ejemplo in test_group["examples"]:
        test_rows.append({
            **ejemplo,
            "group_id": test_group["group_id"],
        })

    # Un grupo adicional para validación en la primera categoría.
    if categoria == list(grupos.keys())[0]:
        for ejemplo in remaining[2]["examples"]:
            validation_rows.append({
                **ejemplo,
                "group_id": remaining[2]["group_id"],
            })

    # Un grupo adicional para prueba en la tercera categoría.
    if categoria == list(grupos.keys())[2]:
        for ejemplo in remaining[2]["examples"]:
            test_rows.append({
                **ejemplo,
                "group_id": remaining[2]["group_id"],
            })

train = pd.DataFrame(train_rows)
validation = pd.DataFrame(validation_rows)
test = pd.DataFrame(test_rows)

# Comprobar que no se comparte ningún grupo entre particiones.
train_groups = set(train["group_id"])
val_groups = set(validation["group_id"])
test_groups = set(test["group_id"])

assert train_groups.isdisjoint(val_groups)
assert train_groups.isdisjoint(test_groups)
assert val_groups.isdisjoint(test_groups)

assert len(train) == 150
assert len(validation) == 30
assert len(test) == 30

os.makedirs("data", exist_ok=True)

train.to_csv("data/train.csv", index=False, encoding="utf-8")
validation.to_csv(
    "data/validation.csv", index=False, encoding="utf-8"
)
test.to_csv("data/test.csv", index=False, encoding="utf-8")

print("Entrenamiento:", len(train))
print("Validación:", len(validation))
print("Prueba:", len(test))

print("\nGrupos originales por partición:")
print("Entrenamiento:", train["group_id"].nunique())
print("Validación:", validation["group_id"].nunique())
print("Prueba:", test["group_id"].nunique())

print("\nCategorías en validación:")
print(validation["category"].value_counts().to_dict())

print("\nCategorías en prueba:")
print(test["category"].value_counts().to_dict())

print("\nArchivos guardados correctamente.")
```

```python

import torch
import json

INSTRUCCIONES_BASE = """
Eres un asistente de clasificación de incidencias de una aseguradora
ficticia de automóviles.

Clasifica la consulta usando exclusivamente estas categorías:
siniestro, poliza, recibos, asistencia, documentacion.

Asigna una prioridad: baja, media o alta.

Devuelve únicamente un objeto JSON válido con exactamente estos campos:
category, priority, response.

category debe ser una de las categorías permitidas.
priority debe ser baja, media o alta.
response debe ser una respuesta breve en español.
No añadas explicaciones fuera del JSON.
"""

def preparar_mensajes(instrucciones, consulta):
    return [
        {"role": "system", "content": instrucciones},
        {"role": "user", "content": consulta},
    ]

def generar_respuesta(modelo, tokenizador, instrucciones, consulta):
    mensajes = preparar_mensajes(instrucciones, consulta)

    texto = tokenizador.apply_chat_template(
        mensajes,
        tokenize=False,
        add_generation_prompt=True,
        enable_thinking=False,
    )

    entradas = tokenizador(
        texto,
        return_tensors="pt",
    ).to(modelo.device)

    with torch.no_grad():
        salida = modelo.generate(
            **entradas,
            max_new_tokens=120,
            do_sample=False,
            pad_token_id=tokenizador.eos_token_id,
        )

    tokens_nuevos = salida[0, entradas["input_ids"].shape[1]:]

    return tokenizador.decode(
        tokens_nuevos,
        skip_special_tokens=True,
    ).strip()

consulta_prueba = (
    "Mi coche se ha averiado en carretera y necesito una grúa."
)

respuesta_base = generar_respuesta(
    model,
    tokenizer,
    INSTRUCCIONES_BASE,
    consulta_prueba,
)

print("CONSULTA:")
print(consulta_prueba)

print("\nRESPUESTA ORIGINAL DEL MODELO:")
print(respuesta_base)

print("\n¿Es JSON válido?")
try:
    objeto = json.loads(respuesta_base)
    print("Sí")
    print(objeto)
except json.JSONDecodeError:
    print("No")
```

```python

import pandas as pd
import torch
from datasets import Dataset
from peft import LoraConfig
from trl import SFTConfig, SFTTrainer

# 1. Preparar cada ejemplo usando la plantilla oficial de Qwen3.
def formatear_ejemplo(fila):
    mensajes = [
        {
            "role": "system",
            "content": INSTRUCCIONES_BASE,
        },
        {
            "role": "user",
            "content": fila["text"],
        },
        {
            "role": "assistant",
            "content": fila["output_json"],
        },
    ]

    return tokenizer.apply_chat_template(
        mensajes,
        tokenize=False,
        add_generation_prompt=False,
        enable_thinking=False,
    )

# Crear una columna con la conversación completa formateada.
train_formateado = train.copy()
validation_formateado = validation.copy()

train_formateado["formatted_text"] = train_formateado.apply(
    formatear_ejemplo, axis=1
)

validation_formateado["formatted_text"] = validation_formateado.apply(
    formatear_ejemplo, axis=1
)

# Pasar a Datasets, conservando solo el texto de entrenamiento.
dataset_train = Dataset.from_pandas(
    train_formateado[["formatted_text"]],
    preserve_index=False,
)

dataset_validation = Dataset.from_pandas(
    validation_formateado[["formatted_text"]],
    preserve_index=False,
)

print("Ejemplos de entrenamiento:", len(dataset_train))
print("Ejemplos de validación:", len(dataset_validation))
print("\nLongitud de ejemplo formateado:",
      len(dataset_train[0]["formatted_text"]))

# 2. Configuración LoRA fija.
lora_config = LoraConfig(
    r=8,
    lora_alpha=16,
    lora_dropout=0.05,
    target_modules=["q_proj", "v_proj"],
    bias="none",
    task_type="CAUSAL_LM",
)

# Evitar problemas de memoria con el modo caché durante el entrenamiento.
model.config.use_cache = False

# 3. Parámetros del entrenamiento.
training_args = SFTConfig(
    output_dir="./resultados_proyecto7",
    num_train_epochs=2,
    per_device_train_batch_size=2,
    per_device_eval_batch_size=2,
    gradient_accumulation_steps=4,
    learning_rate=2e-4,
    logging_steps=5,
    eval_strategy="epoch",
    save_strategy="epoch",
    save_total_limit=1,
    fp16=True,
    max_length=512,
    dataset_text_field="formatted_text",
    packing=False,
    report_to="none",
    seed=42,
    gradient_checkpointing=False,
)

# 4. Crear el entrenador.
trainer = SFTTrainer(
    model=model,
    args=training_args,
    train_dataset=dataset_train,
    eval_dataset=dataset_validation,
    processing_class=tokenizer,
    peft_config=lora_config,
)

# 5. Entrenar.
resultado_entrenamiento = trainer.train()

print("\nEntrenamiento terminado.")
print("Métricas:", resultado_entrenamiento.metrics)

# Guardar el adaptador LoRA y el tokenizador.
trainer.model.save_pretrained("./adaptador_proyecto7")
tokenizer.save_pretrained("./adaptador_proyecto7")

print("\nAdaptador guardado en ./adaptador_proyecto7")
```

```python

from datasets import Dataset

def formatear_ejemplo(fila):
    mensajes = [
        {
            "role": "system",
            "content": INSTRUCCIONES_BASE
        },
        {
            "role": "user",
            "content": fila["text"]
        },
        {
            "role": "assistant",
            "content": fila["output_json"]
        }
    ]

    return tokenizer.apply_chat_template(
        mensajes,
        tokenize=False,
        add_generation_prompt=False,
        enable_thinking=False
    )

train_formateado = train.copy()
validation_formateado = validation.copy()

train_formateado["formatted_text"] = train_formateado.apply(
    formatear_ejemplo,
    axis=1
)

validation_formateado["formatted_text"] = validation_formateado.apply(
    formatear_ejemplo,
    axis=1
)

dataset_train = Dataset.from_pandas(
    train_formateado[["formatted_text"]],
    preserve_index=False
)

dataset_validation = Dataset.from_pandas(
    validation_formateado[["formatted_text"]],
    preserve_index=False
)

print("Ejemplos preparados para entrenar:", len(dataset_train))
print("Ejemplos preparados para validar:", len(dataset_validation))
print("\nPrimer ejemplo formateado:\n")
print(dataset_train[0]["formatted_text"][:1000])
```

```python

import torch
from peft import LoraConfig
from trl import SFTConfig, SFTTrainer

# 1. Configuración de LoRA
lora_config = LoraConfig(
    r=8,
    lora_alpha=16,
    lora_dropout=0.05,
    target_modules=["q_proj", "v_proj"],
    bias="none",
    task_type="CAUSAL_LM",
)

# 2. Preparar el modelo para el entrenamiento
model.config.use_cache = False

# 3. Configuración del entrenamiento
training_args = SFTConfig(
    output_dir="./resultados_proyecto7",
    num_train_epochs=2,
    per_device_train_batch_size=2,
    per_device_eval_batch_size=2,
    gradient_accumulation_steps=4,
    learning_rate=2e-4,
    logging_steps=5,
    eval_strategy="epoch",
    save_strategy="epoch",
    save_total_limit=1,
    fp16=True,
    max_length=512,
    dataset_text_field="formatted_text",
    packing=False,
    report_to="none",
    seed=42,
    gradient_checkpointing=False,
)

# 4. Crear el entrenador
trainer = SFTTrainer(
    model=model,
    args=training_args,
    train_dataset=dataset_train,
    eval_dataset=dataset_validation,
    processing_class=tokenizer,
    peft_config=lora_config,
)

print("Entrenador creado correctamente.")
print("Configuración LoRA:")
print("r =", lora_config.r)
print("alpha =", lora_config.lora_alpha)
print("dropout =", lora_config.lora_dropout)

# 5. Entrenar
resultado_entrenamiento = trainer.train()

print("\nEntrenamiento terminado.")
print("Métricas del entrenamiento:")
print(resultado_entrenamiento.metrics)

# 6. Guardar el adaptador LoRA
trainer.model.save_pretrained("./adaptador_proyecto7")
tokenizer.save_pretrained("./adaptador_proyecto7")

print("\nAdaptador guardado en ./adaptador_proyecto7")
```

```python

import torch
import json

def generar_respuesta(modelo, consulta, usar_few_shot=False):
    mensajes = [
        {
            "role": "system",
            "content": INSTRUCCIONES_BASE
        }
    ]

    # Añadir los cinco ejemplos si estamos evaluando few-shot
    if usar_few_shot:
        mensajes.extend(ejemplos_few_shot)

    # Añadir la consulta que queremos resolver
    mensajes.append({
        "role": "user",
        "content": consulta
    })

    # Aplicar la plantilla oficial de conversación de Qwen
    texto = tokenizer.apply_chat_template(
        mensajes,
        tokenize=False,
        add_generation_prompt=True,
        enable_thinking=False
    )

    entradas = tokenizer(
        texto,
        return_tensors="pt"
    ).to(modelo.device)

    with torch.no_grad():
        salida = modelo.generate(
            **entradas,
            max_new_tokens=120,
            do_sample=False,
            pad_token_id=tokenizer.pad_token_id,
            eos_token_id=tokenizer.eos_token_id
        )

    # Conservar únicamente los tokens generados como respuesta
    tokens_nuevos = salida[0, entradas["input_ids"].shape[1]:]

    respuesta = tokenizer.decode(
        tokens_nuevos,
        skip_special_tokens=True
    ).strip()

    return respuesta


print("Función de generación creada correctamente.")
```

```python

import json
import pandas as pd
from IPython.display import display

CATEGORIAS_VALIDAS = {
    "siniestro",
    "poliza",
    "recibos",
    "asistencia",
    "documentacion",
}

PRIORIDADES_VALIDAS = {"baja", "media", "alta"}

resultados_validacion = []

for indice, fila in validation.iterrows():
    consulta = fila["text"]
    categoria_esperada = fila["category"]
    prioridad_esperada = fila["priority"]

    configuraciones = [
        ("base", modelo_base_limpio, False),
        ("few_shot", modelo_base_limpio, True),
        ("lora", modelo_lora, False),
    ]

    for nombre, modelo_actual, usar_few_shot in configuraciones:
        respuesta = generar_respuesta(
            modelo_actual,
            consulta,
            usar_few_shot=usar_few_shot,
        )

        # Validación estricta: no se corrige ni se repara la respuesta
        json_valido = False
        categoria_obtenida = None
        prioridad_obtenida = None

        try:
            objeto = json.loads(respuesta)

            if (
                isinstance(objeto, dict)
                and set(objeto.keys()) == {
                    "category", "priority", "response"
                }
                and isinstance(objeto["category"], str)
                and isinstance(objeto["priority"], str)
                and isinstance(objeto["response"], str)
                and objeto["category"] in CATEGORIAS_VALIDAS
                and objeto["priority"] in PRIORIDADES_VALIDAS
            ):
                json_valido = True
                categoria_obtenida = objeto["category"]
                prioridad_obtenida = objeto["priority"]

        except (json.JSONDecodeError, TypeError):
            pass

        resultados_validacion.append({
            "id": int(indice),
            "modelo": nombre,
            "consulta": consulta,
            "categoria_esperada": categoria_esperada,
            "prioridad_esperada": prioridad_esperada,
            "respuesta_original": respuesta,
            "json_valido": json_valido,
            "categoria_obtenida": categoria_obtenida,
            "prioridad_obtenida": prioridad_obtenida,
            "categoria_correcta": (
                json_valido
                and categoria_obtenida == categoria_esperada
            ),
            "prioridad_correcta": (
                json_valido
                and prioridad_obtenida == prioridad_esperada
            ),
        })

    print(
        f"Consulta {len(resultados_validacion) // 3}/"
        f"{len(validation)} evaluada"
    )

df_validacion = pd.DataFrame(resultados_validacion)

resumen_validacion = (
    df_validacion.groupby("modelo")
    .agg(
        casos=("id", "count"),
        json_validos=("json_valido", "sum"),
        categorias_correctas=("categoria_correcta", "sum"),
        prioridades_correctas=("prioridad_correcta", "sum"),
    )
    .reset_index()
)

for metrica, columna in [
    ("JSON válido (%)", "json_validos"),
    ("Categoría correcta (%)", "categorias_correctas"),
    ("Prioridad correcta (%)", "prioridades_correctas"),
]:
    resumen_validacion[metrica] = (
        resumen_validacion[columna] / resumen_validacion["casos"] * 100
    ).round(2)

print("\n=== RESUMEN DE VALIDACIÓN ===")
display(resumen_validacion)

# Guardar las respuestas originales para documentar los resultados
df_validacion.to_csv(
    "/content/resultados_proyecto7/resultados_validacion.csv",
    index=False,
    encoding="utf-8-sig",
)

resumen_validacion.to_csv(
    "/content/resultados_proyecto7/resumen_validacion.csv",
    index=False,
    encoding="utf-8-sig",
)

print("\nResultados guardados en /content/resultados_proyecto7/")
```

```python

# ============================================================
# PROYECTO 7 — MEJORA DE LOS EJEMPLOS FEW-SHOT
# Objetivo: cubrir categorías y prioridades diferentes
# Se utiliza exclusivamente el conjunto de entrenamiento
# ============================================================

import pandas as pd

# Trabajamos con una copia del conjunto de entrenamiento
datos_ejemplos = train.copy().reset_index(drop=True)

categorias_objetivo = {
    "siniestro",
    "poliza",
    "recibos",
    "asistencia",
    "documentacion",
}

prioridades_objetivo = {"baja", "media", "alta"}

# Cada ejemplo puede cubrir dos características:
# su categoría y su prioridad.
caracteristicas_pendientes = {
    ("categoria", categoria)
    for categoria in categorias_objetivo
} | {
    ("prioridad", prioridad)
    for prioridad in prioridades_objetivo
}

seleccionados = []
indices_disponibles = set(datos_ejemplos.index)

# Selección progresiva: priorizamos ejemplos que cubran
# características que todavía no estén representadas.
while caracteristicas_pendientes and indices_disponibles:
    mejor_indice = None
    mejor_puntuacion = -1

    for indice in indices_disponibles:
        fila = datos_ejemplos.loc[indice]

        caracteristicas = {
            ("categoria", fila["category"]),
            ("prioridad", fila["priority"]),
        }

        puntuacion = len(
            caracteristicas & caracteristicas_pendientes
        )

        if puntuacion > mejor_puntuacion:
            mejor_puntuacion = puntuacion
            mejor_indice = indice

    if mejor_indice is None or mejor_puntuacion == 0:
        break

    fila = datos_ejemplos.loc[mejor_indice]
    seleccionados.append(fila)

    caracteristicas_ejemplo = {
        ("categoria", fila["category"]),
        ("prioridad", fila["priority"]),
    }

    caracteristicas_pendientes -= caracteristicas_ejemplo
    indices_disponibles.remove(mejor_indice)

# Convertimos la selección en una tabla
few_shot_mejorado = pd.DataFrame(seleccionados)

print("NÚMERO DE EJEMPLOS SELECCIONADOS:", len(few_shot_mejorado))
print("\nDISTRIBUCIÓN POR CATEGORÍA:")
print(few_shot_mejorado["category"].value_counts())

print("\nDISTRIBUCIÓN POR PRIORIDAD:")
print(few_shot_mejorado["priority"].value_counts())

print("\nEJEMPLOS SELECCIONADOS:")
for _, fila in few_shot_mejorado.iterrows():
    print("\nCategoría:", fila["category"])
    print("Prioridad:", fila["priority"])
    print("Consulta:", fila["text"])
    print("Respuesta:", fila["output_json"])

# Creamos los mensajes que utilizará el prompt few-shot
ejemplos_few_shot = []

for _, fila in few_shot_mejorado.iterrows():
    ejemplos_few_shot.extend([
        {"role": "user", "content": fila["text"]},
        {"role": "assistant", "content": fila["output_json"]},
    ])

print("\nEjemplos few-shot preparados correctamente.")
```

```python

# ============================================================
# PROYECTO 7: AUDITORÍA DE LAS ETIQUETAS DE PRIORIDAD
# No modifica los datos originales ni entrena el modelo.
# ============================================================

import pandas as pd
from IPython.display import display

def sugerir_prioridad(texto, categoria):
    """
    Propuesta orientativa basada en reglas sencillas.
    No sustituye la revisión humana de las etiquetas.
    """
    texto = str(texto).lower()
    categoria = str(categoria).lower()

    # Alta: urgencia o necesidad de actuación inmediata
    indicadores_alta = [
        "urgente",
        "urgencia",
        "inmediata",
        "inmediatamente",
        "accidente",
        "averiado",
        "grúa",
        "no reconozco",
        "fraude",
        "bloqueado",
        "en carretera",
    ]

    if any(palabra in texto for palabra in indicadores_alta):
        return "alta"

    # Media: gestión activa o seguimiento de una incidencia
    indicadores_media = [
        "siniestro",
        "expediente",
        "daños",
        "parte",
        "modificar",
        "cambiar",
        "añadir",
        "actualizar",
        "faltan",
        "pendientes",
        "asistencia",
        "traslado",
        "vehículo de sustitución",
        "cobertura",
    ]

    if any(palabra in texto for palabra in indicadores_media):
        return "media"

    # Baja: consultas ordinarias de información o documentación
    return "baja"


# Revisamos las consultas originales de entrenamiento y validación.
# El conjunto de prueba permanece intacto y fuera de esta auditoría.
for nombre, datos in [
    ("ENTRENAMIENTO", train),
    ("VALIDACIÓN", validation),
]:
    originales = (
        datos[["group_id", "text", "category", "priority"]]
        .drop_duplicates(subset="group_id")
        .copy()
    )

    originales["prioridad_sugerida"] = originales.apply(
        lambda fila: sugerir_prioridad(
            fila["text"], fila["category"]
        ),
        axis=1,
    )

    originales["coincide_etiqueta_actual"] = (
        originales["priority"]
        == originales["prioridad_sugerida"]
    )

    print("\n" + "=" * 75)
    print(f"AUDITORÍA: {nombre}")
    display(
        originales[
            [
                "group_id",
                "text",
                "category",
                "priority",
                "prioridad_sugerida",
                "coincide_etiqueta_actual",
            ]
        ].sort_values(["category", "group_id"])
    )

    print("\nResumen de coincidencias:")
    print(
        originales["coincide_etiqueta_actual"]
        .value_counts()
        .rename(index={True: "Coincide", False: "No coincide"})
    )
```

```python

# PROYECTO 7: NUEVO ENTRENAMIENTO LoRA DESDE CERO

import os
import gc
import json
import torch

from transformers import AutoTokenizer, AutoModelForCausalLM
from peft import LoraConfig
from trl import SFTConfig, SFTTrainer

# 1. Comprobar que los datasets están preparados
assert len(dataset_train_nuevo) == 150
assert len(dataset_validation_nuevo) == 30

# 2. Liberar modelos y entrenadores anteriores para evitar
# reutilizar accidentalmente el adaptador antiguo
for nombre in [
    "trainer", "modelo_lora", "modelo_base",
    "modelo_base_limpio", "model"
]:
    if nombre in globals():
        del globals()[nombre]

gc.collect()

if torch.cuda.is_available():
    torch.cuda.empty_cache()

# 3. Cargar un tokenizer y un modelo base nuevos
OPEN_MODEL_ID = "Qwen/Qwen3-0.6B"

tokenizer_nuevo = AutoTokenizer.from_pretrained(OPEN_MODEL_ID)

modelo_nuevo = AutoModelForCausalLM.from_pretrained(
    OPEN_MODEL_ID,
    dtype=torch.float16,
    device_map="auto"
)

# Desactivar caché durante el entrenamiento
modelo_nuevo.config.use_cache = False

# 4. Configurar LoRA
lora_config_nuevo = LoraConfig(
    r=8,
    lora_alpha=16,
    lora_dropout=0.05,
    target_modules=["q_proj", "v_proj"],
    bias="none",
    task_type="CAUSAL_LM"
)

# 5. Crear una carpeta nueva para no mezclar resultados anteriores
ruta_salida = "/content/resultados_proyecto7_v2"
ruta_adaptador = "/content/adaptador_proyecto7_v2"

os.makedirs(ruta_salida, exist_ok=True)

# 6. Configurar el entrenamiento
training_args_nuevo = SFTConfig(
    output_dir=ruta_salida,
    num_train_epochs=2,
    per_device_train_batch_size=2,
    per_device_eval_batch_size=2,
    gradient_accumulation_steps=4,
    learning_rate=2e-4,
    logging_steps=5,
    eval_strategy="epoch",
    save_strategy="epoch",
    save_total_limit=1,
    fp16=True,
    max_length=512,
    dataset_text_field="formatted_text",
    packing=False,
    report_to="none",
    seed=42,
    gradient_checkpointing=False
)

# 7. Crear el entrenador con los datasets actualizados
trainer_nuevo = SFTTrainer(
    model=modelo_nuevo,
    args=training_args_nuevo,
    train_dataset=dataset_train_nuevo,
    eval_dataset=dataset_validation_nuevo,
    processing_class=tokenizer_nuevo,
    peft_config=lora_config_nuevo
)

# 8. Entrenar
resultado_entrenamiento_nuevo = trainer_nuevo.train()

# 9. Guardar el adaptador nuevo y el tokenizer
trainer_nuevo.model.save_pretrained(ruta_adaptador)
tokenizer_nuevo.save_pretrained(ruta_adaptador)

# 10. Guardar las métricas del entrenamiento
metricas = resultado_entrenamiento_nuevo.metrics

metricas["modelo_base"] = OPEN_MODEL_ID
metricas["ejemplos_entrenamiento"] = len(dataset_train_nuevo)
metricas["ejemplos_validacion"] = len(dataset_validation_nuevo)
metricas["epocas"] = 2
metricas["learning_rate"] = 2e-4
metricas["lora_r"] = 8
metricas["lora_alpha"] = 16
metricas["prompt_actualizado"] = True
metricas["etiquetas_corregidas"] = True

ruta_metricas = os.path.join(
    ruta_salida,
    "metricas_entrenamiento_v2.json"
)

with open(ruta_metricas, "w", encoding="utf-8") as archivo:
    json.dump(
        metricas,
        archivo,
        ensure_ascii=False,
        indent=4,
        default=str
    )

print("\nENTRENAMIENTO FINALIZADO")
print("Adaptador guardado en:", ruta_adaptador)
print("Métricas guardadas en:", ruta_metricas)
print("\nMétricas del entrenamiento:")
for clave, valor in metricas.items():
    print(f"{clave}: {valor}")
```

```python

# PROYECTO 7: EVALUACIÓN COMPARATIVA EN VALIDACIÓN

import json
import os
import pandas as pd
from IPython.display import display

# Comprobar que existen los objetos necesarios
assert len(validation) == 30
assert "modelo_base_eval" in globals()
assert "modelo_lora_v2" in globals()
assert "ejemplos_few_shot_v2" in globals()

# Validar la salida estrictamente, sin reparar el JSON
CATEGORIAS_VALIDAS = {
    "siniestro", "poliza", "recibos",
    "asistencia", "documentacion"
}
PRIORIDADES_VALIDAS = {"baja", "media", "alta"}
CLAVES_ESPERADAS = {"category", "priority", "response"}

def validar_salida_estricta(texto):
    try:
        objeto = json.loads(texto)
    except Exception:
        return {
            "json_valido": False,
            "category_pred": None,
            "priority_pred": None,
            "response_pred": None
        }

    if not isinstance(objeto, dict):
        return {
            "json_valido": False,
            "category_pred": None,
            "priority_pred": None,
            "response_pred": None
        }

    if set(objeto.keys()) != CLAVES_ESPERADAS:
        return {
            "json_valido": False,
            "category_pred": None,
            "priority_pred": None,
            "response_pred": None
        }

    if objeto["category"] not in CATEGORIAS_VALIDAS:
        return {
            "json_valido": False,
            "category_pred": None,
            "priority_pred": None,
            "response_pred": None
        }

    if objeto["priority"] not in PRIORIDADES_VALIDAS:
        return {
            "json_valido": False,
            "category_pred": None,
            "priority_pred": None,
            "response_pred": None
        }

    if not isinstance(objeto["response"], str):
        return {
            "json_valido": False,
            "category_pred": None,
            "priority_pred": None,
            "response_pred": None
        }

    return {
        "json_valido": True,
        "category_pred": objeto["category"],
        "priority_pred": objeto["priority"],
        "response_pred": objeto["response"]
    }

# Generar predicciones sobre las mismas 30 consultas
resultados_validacion_v2 = []

for indice, fila in validation.reset_index(drop=True).iterrows():
    consulta = fila["text"]

    predicciones = {
        "base": generar_respuesta_v2(
            modelo_base_eval, consulta, usar_few_shot=False
        ),
        "few_shot": generar_respuesta_v2(
            modelo_base_eval, consulta, usar_few_shot=True
        ),
        "lora_v2": generar_respuesta_v2(
            modelo_lora_v2, consulta, usar_few_shot=False
        )
    }

    for nombre_modelo, salida_cruda in predicciones.items():
        evaluacion = validar_salida_estricta(salida_cruda)

        resultados_validacion_v2.append({
            "indice": indice,
            "group_id": fila["group_id"],
            "text": consulta,
            "category_real": fila["category"],
            "priority_real": fila["priority"],
            "modelo": nombre_modelo,
            "salida_cruda": salida_cruda,
            **evaluacion,
            "categoria_correcta": (
                evaluacion["category_pred"] == fila["category"]
                if evaluacion["json_valido"] else False
            ),
            "prioridad_correcta": (
                evaluacion["priority_pred"] == fila["priority"]
                if evaluacion["json_valido"] else False
            )
        })

    if (indice + 1) % 5 == 0:
        print(f"Consultas procesadas: {indice + 1}/30")

df_validacion_v2 = pd.DataFrame(resultados_validacion_v2)

# Resumen por modelo
resumen_validacion_v2 = (
    df_validacion_v2.groupby("modelo")
    .agg(
        ejemplos=("indice", "count"),
        json_validos=("json_valido", "sum"),
        categorias_correctas=("categoria_correcta", "sum"),
        prioridades_correctas=("prioridad_correcta", "sum")
    )
    .reset_index()
)

resumen_validacion_v2["json_valido_pct"] = (
    100 * resumen_validacion_v2["json_validos"]
    / resumen_validacion_v2["ejemplos"]
).round(2)

resumen_validacion_v2["categoria_correcta_pct"] = (
    100 * resumen_validacion_v2["categorias_correctas"]
    / resumen_validacion_v2["ejemplos"]
).round(2)

resumen_validacion_v2["prioridad_correcta_pct"] = (
    100 * resumen_validacion_v2["prioridades_correctas"]
    / resumen_validacion_v2["ejemplos"]
).round(2)

# Guardar resultados sin modificar las salidas originales
os.makedirs("/content/resultados_proyecto7_v2", exist_ok=True)

df_validacion_v2.to_csv(
    "/content/resultados_proyecto7_v2/"
    "predicciones_validacion_v2.csv",
    index=False,
    encoding="utf-8-sig"
)

resumen_validacion_v2.to_csv(
    "/content/resultados_proyecto7_v2/"
    "resumen_validacion_v2.csv",
    index=False,
    encoding="utf-8-sig"
)

print("\nRESUMEN DE VALIDACIÓN")
display(resumen_validacion_v2)

print("\nARCHIVOS GUARDADOS")
print("/content/resultados_proyecto7_v2/predicciones_validacion_v2.csv")
print("/content/resultados_proyecto7_v2/resumen_validacion_v2.csv")
```

```python

import json
import pandas as pd
from pathlib import Path

# 1. Cargar la validación con la regla de negocio corregida
ruta_validacion = Path(
    "/content/resultados_proyecto7_v3/validation_corregida_v3.csv"
)

assert ruta_validacion.exists(), (
    f"No se encuentra el archivo: {ruta_validacion}"
)

validacion_v3 = pd.read_csv(ruta_validacion)

assert len(validacion_v3) == 30, (
    f"Se esperaban 30 ejemplos y hay {len(validacion_v3)}."
)

CATEGORIAS_VALIDAS = {
    "siniestro", "poliza", "recibos",
    "asistencia", "documentacion"
}
PRIORIDADES_VALIDAS = {"baja", "media", "alta"}
CLAVES_ESPERADAS = {"category", "priority", "response"}


# 2. Validador estricto: no elimina Markdown ni repara el JSON
def validar_salida_estricta_v3(salida):
    resultado = {
        "json_valido": False,
        "category_pred": None,
        "priority_pred": None,
        "response_pred": None
    }

    try:
        objeto = json.loads(salida)

        if not isinstance(objeto, dict):
            return resultado

        if set(objeto.keys()) != CLAVES_ESPERADAS:
            return resultado

        if objeto["category"] not in CATEGORIAS_VALIDAS:
            return resultado

        if objeto["priority"] not in PRIORIDADES_VALIDAS:
            return resultado

        if not isinstance(objeto["response"], str):
            return resultado

        resultado.update({
            "json_valido": True,
            "category_pred": objeto["category"],
            "priority_pred": objeto["priority"],
            "response_pred": objeto["response"]
        })

        return resultado

    except (json.JSONDecodeError, TypeError):
        return resultado


# 3. Evaluar los 30 ejemplos con el mismo modelo y prompt V3
resultados = []

for indice, fila in validacion_v3.iterrows():
    consulta = str(fila["text"])

    salida = generar_respuesta_v3(
        modelo_base_eval,
        consulta,
        usar_few_shot=True
    )

    validacion_salida = validar_salida_estricta_v3(salida)

    categoria_real = fila["category"]
    prioridad_real = fila["priority"]

    resultados.append({
        "indice": indice,
        "group_id": fila["group_id"],
        "text": consulta,
        "category_real": categoria_real,
        "priority_real": prioridad_real,
        "modelo": "base_few_shot_v3",
        "salida_cruda": salida,
        **validacion_salida,
        "categoria_correcta": (
            validacion_salida["json_valido"]
            and validacion_salida["category_pred"] == categoria_real
        ),
        "prioridad_correcta": (
            validacion_salida["json_valido"]
            and validacion_salida["priority_pred"] == prioridad_real
        )
    })

    print(f"Evaluados: {len(resultados)}/30")

df_validacion_v3 = pd.DataFrame(resultados)


# 4. Calcular métricas
resumen_v3 = pd.DataFrame([{
    "modelo": "Base + few-shot V3",
    "ejemplos": len(df_validacion_v3),
    "json_validos": int(df_validacion_v3["json_valido"].sum()),
    "categorias_correctas": int(
        df_validacion_v3["categoria_correcta"].sum()
    ),
    "prioridades_correctas": int(
        df_validacion_v3["prioridad_correcta"].sum()
    ),
    "json_pct": round(
        100 * df_validacion_v3["json_valido"].mean(), 2
    ),
    "categoria_pct": round(
        100 * df_validacion_v3["categoria_correcta"].mean(), 2
    ),
    "prioridad_pct": round(
        100 * df_validacion_v3["prioridad_correcta"].mean(), 2
    )
}])

# 5. Guardar resultados completos y resumen
carpeta_salida = Path("/content/resultados_proyecto7_v3")
carpeta_salida.mkdir(parents=True, exist_ok=True)

df_validacion_v3.to_csv(
    carpeta_salida / "predicciones_validacion_v3.csv",
    index=False,
    encoding="utf-8-sig"
)

resumen_v3.to_csv(
    carpeta_salida / "resumen_validacion_v3.csv",
    index=False,
    encoding="utf-8-sig"
)

print("\n" + "=" * 70)
print("RESUMEN DE VALIDACIÓN V3")
print("=" * 70)
print(resumen_v3.to_string(index=False))

print("\nResultados guardados en:")
print(carpeta_salida / "predicciones_validacion_v3.csv")
print(carpeta_salida / "resumen_validacion_v3.csv")

print("\nPrimeros errores de clasificación o formato:")
errores = df_validacion_v3[
    (~df_validacion_v3["json_valido"])
    | (~df_validacion_v3["categoria_correcta"])
    | (~df_validacion_v3["prioridad_correcta"])
]

if errores.empty:
    print("No se han encontrado errores.")
else:
    print(
        errores[
            [
                "group_id", "text", "category_real", "priority_real",
                "salida_cruda", "json_valido",
                "categoria_correcta", "prioridad_correcta"
            ]
        ].head(15).to_string(index=False)
    )
```

```python

import torch

def generar_respuesta_final(modelo, consulta, configuracion):

    if configuracion == "base_basico":
        prompt = prompt_basico_final
        usar_few_shot = False

    elif configuracion == "base_few_shot":
        prompt = prompt_v3_final
        usar_few_shot = True

    elif configuracion == "lora_v3":
        prompt = prompt_v3_final
        usar_few_shot = False

    else:
        raise ValueError(
            f"Configuración desconocida: {configuracion}"
        )

    mensajes = [
        {"role": "system", "content": prompt}
    ]

    if usar_few_shot:
        mensajes.extend(ejemplos_few_shot_v2)

    mensajes.append({
        "role": "user",
        "content": str(consulta)
    })

    # Devolver un BatchEncoding con input_ids y attention_mask.
    entradas = tokenizer_eval.apply_chat_template(
        mensajes,
        tokenize=True,
        add_generation_prompt=True,
        enable_thinking=False,
        return_tensors="pt",
        return_dict=True
    )

    dispositivo = next(modelo.parameters()).device
    entradas = {
        clave: valor.to(dispositivo)
        for clave, valor in entradas.items()
        if isinstance(valor, torch.Tensor)
    }

    modelo.eval()

    with torch.inference_mode():
        salida = modelo.generate(
            **entradas,
            max_new_tokens=120,
            do_sample=False,
            pad_token_id=tokenizer_eval.eos_token_id
        )

    # Obtener la longitud a partir de input_ids.
    longitud_entrada = entradas["input_ids"].shape[1]
    tokens_nuevos = salida[0][longitud_entrada:]

    respuesta = tokenizer_eval.decode(
        tokens_nuevos,
        skip_special_tokens=True
    )

    return respuesta.strip()


# Prueba técnica con una sola consulta de validación.
prueba = generar_respuesta_final(
    modelo_base_eval,
    "Quiero descargar un recibo de mi póliza.",
    "base_basico"
)

print("Prueba ejecutada correctamente.")
print("Respuesta original del modelo:")
print(prueba)
```

```python

import os
import pandas as pd

rutas = {
    "entrenamiento": "/content/resultados_proyecto7_v3/train_corregido_v3.csv",
    "validacion": "/content/resultados_proyecto7_v3/validation_final_v3.csv",
    "test": "/content/resultados_proyecto7_v3/test_corregido_v3.csv"
}

datos = {}

for nombre, ruta in rutas.items():
    if not os.path.exists(ruta):
        print(f"NO ENCONTRADO: {ruta}")
        continue

    df = pd.read_csv(ruta)
    datos[nombre] = df

    print("\n" + "=" * 55)
    print("PARTICIÓN:", nombre.upper())
    print("Filas:", len(df))
    print("Columnas:", list(df.columns))

    if "group_id" in df.columns:
        print("Grupos únicos:", df["group_id"].nunique())
        print("Grupos nulos:", df["group_id"].isna().sum())
    else:
        print("AVISO: no hay columna group_id.")

    if "text" in df.columns:
        print("Textos duplicados:", df["text"].duplicated().sum())

print("\n" + "=" * 55)
print("COMPROBACIÓN DE GRUPOS COMPARTIDOS")

if len(datos) == 3 and all(
    "group_id" in df.columns for df in datos.values()
):
    grupos = {
        nombre: set(df["group_id"].dropna().astype(str))
        for nombre, df in datos.items()
    }

    pares = [
        ("entrenamiento", "validacion"),
        ("entrenamiento", "test"),
        ("validacion", "test")
    ]

    for a, b in pares:
        comunes = grupos[a] & grupos[b]
        print(f"{a} / {b}: {len(comunes)} grupos compartidos")
        if comunes:
            print("  Ejemplos:", sorted(comunes)[:10])
else:
    print("No se pudo completar: falta alguna partición o group_id.")

print("\nAuditoría terminada. No se ha modificado ningún archivo.")
```
